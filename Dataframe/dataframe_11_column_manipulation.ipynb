```python
from pyspark.sql import SparkSession
import getpass
```

```python
username = getpass.getuser()
```

```python
spark = SparkSession.builder. \
config('spark.ui.port','0'). \
config("spark.sql.warehouse.dir", f"/user/{username}/warehouse"). \
enableHiveSupport(). \
master('yarn'). \
getOrCreate()
```

```python
load_df = spark.read.format('csv').option('inferschema', 'true').load('datasets/retail_db/products/part-00000')
```

```python
load_df.show(10)
```

```python
load_df.printSchema()
```

```python
product_columns = ["product_id", "product_category_id", "product_name", "product_description", "product_price", "product_image"]
```

```python
product_df = load_df.toDF(*product_columns)
```

```python
product_df.show(10)
```

```python
from pyspark.sql.functions import col
twice_product_df = product_df. \
select("product_id", "product_category_id", "product_name", "product_description", (col("product_price")*2).alias("twice_product_price"), "product_image")
```

```python
twice_product_df.show()
```
