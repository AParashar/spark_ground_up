```python
from pyspark.sql import SparkSession
import getpass

username = getpass.getuser()

spark = SparkSession.builder \
.config("spark.sql.warehouse/dir", f"/user/{username}/warehouse") \
.config("spark.ui.port", '0') \
.enableHiveSupport() \
.master("yarn") \
.appName("0253") \
.getOrCreate()
```

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, FloatType
product_schema = StructType([
    StructField("ProductID", IntegerType()),
    StructField("Category", StringType()),
    StructField("ProductName", StringType()),
    StructField("Description", StringType()),
    StructField("Price", FloatType()),
    StructField("ImageURL", StringType())
])
product_df = spark.read.format("csv").option("header", False).schema(product_schema).load("datasets/retail_db_product_part-00000")
```

### 3. Find the top 5 most expensive products based on their price, along with their product name, category, and image URL

```python
display(product_df)
```

```python
product_df.select("productName", "category","price", "imageUrl").orderBy("price", ascending=False).limit(5)
```

```python
product_df.createOrReplaceTempView("product")
```

```python
spark.sql("select productName, category, price, imageUrl from product order by price desc limit 5")
```
