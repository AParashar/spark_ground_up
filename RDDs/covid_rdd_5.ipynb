```python
from pyspark.sql import SparkSession
import getpass
username = getpass.getuser()

spark = SparkSession. \
builder. \
config('spark.ui.port','0'). \
config("spark.sql.warehouse.dir", f"/user/{username}/warehouse"). \
enableHiveSupport(). \
master('yarn'). \
getOrCreate()
```

```python
sc = spark.sparkContext
covid_cases_rdd = sc.textFile("data/datasets/covid19/cases/covid_dataset_cases.csv")
covid_states_rdd = sc.textFile("data/datasets/covid19/states/covid_dataset_states.csv")
```

### 5. Find the total number of people hospitalized currently.

```python
# hospitalizedCurrently
covid_cases_col = covid_cases_rdd.map(lambda x: int(x.split(",")[5]))
```

```python
covid_cases_col.reduce(lambda x,y: x+y)
```
